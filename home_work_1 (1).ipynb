```python
#задание 3
low = int(input("Нижняя граница: "))
high = int(input("Верхняя граница: "))

while low <= high:
    mid = (low + high) // 2
    ans = input(f"Число равно {mid}?: ")
    if ans.lower() == 'да':
        print("Ответ: ", mid)
        break
    ans2 = input(f"Число меньше {mid}?: ")
    if ans2.lower() == 'да':
        high = mid - 1
    else:
        low = mid + 1
```

```python
#задание 6
import random

a = [random.randint(0, 100) for _ in range(10)]
print("исходный массив:", a)

s = sorted(a)
n = len(s)

if n % 2 == 1:
    med = s[n // 2]
else:
    med = (s[n // 2 - 1] + s[n // 2]) / 2

print("min =", min(a))
print("max =", max(a))
print("median =", med)
```

```python
#задание 7
a = [random.randint(0, 100) for _ in range(20)]
print("исходный список:", a)

bins = [0] * 10

for x in a:
    idx = min(x // 10, 9)   # 100 попадает в последнюю корзину
    bins[idx] += 1

print("гистограмма:", bins)

probs = [c / len(a) for c in bins]
print("вероятности:", probs)
```

```python
#задание 8
n = int(input("n = "))
scalar = float(input("scalar = "))

a = [random.random() for i in range(n)]
b = [random.random() for i in range(n)]

print("исходный вектор a:", a)
print("исходный вектор b:", b)

sumv = [a[i] + b[i] for i in range(n)]
mulv = [a[i] * b[i] for i in range(n)]

print("покомпонентная сумма:", sumv)
print("покомпонентное умножение:", mulv)

norma = sum(x*x for x in a) ** 0.5
normb = sum(x*x for x in b) ** 0.5

if norma > normb:
    print("норма a больше, умножение на скаляр:", [x * scalar for x in a])
else:
    print("норма b больше, умножение на скаляр:", [x * scalar for x in b])
```

```python
#задание 9
m = int(input("m = "))
n = int(input("n = "))

A = [[random.random() for i in range(n)] for i in range(m)]
v = [random.random() for i in range(n)]

print("матрица:")
for row in A:
    print(' '.join(f"{x:.2f}" for x in row))

print("Матрица A:")
for row in A:
    print(' '.join(f"{x:.2f}" for x in row))

print("Вектор v:")
print(' '.join(f"{x:.2f}" for x in v))

res = []
for i in range(m):
    s = 0
    for j in range(n):
        s += A[i][j] * v[j]
    res.append(s)

print(res)
```

```python
#задание 10
a = [random.randint(-10, 10) for i in range(10)]
orig = a[:]

print("исходный список:", a)

for i in range(len(a)):
    if a[i] < 0:
        left = None
        right = None

        for j in range(i - 1, -1, -1):
            if orig[j] >= 0:
                left = orig[j]
                break

        for j in range(i + 1, len(orig)):
            if orig[j] >= 0:
                right = orig[j]
                break

        if left is not None and right is not None:
            a[i] = (left + right) / 2
        else:
            if left is None:
                left = 0
            if right is None:
                right = 0
            a[i] = (left + right) / 2

print("список после замены отрицательных:", a)
```

```python
#задание 11
data = [random.randint(0, 9) for i in range(10)]
kernel = [random.randint(-2, 2) for i in range(3)]

print("исходный массив данных:", data)
print("ядро свертки:", kernel)

res = []
for i in range(len(data) - len(kernel) + 1):
    s = 0
    for j in range(len(kernel)):
        s += data[i + j] * kernel[j]
    res.append(s)

print("результат свёртки:", res)
```
