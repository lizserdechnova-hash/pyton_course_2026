```python
def table_sum():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}+{j} = {i+j}")
        print()

def table_sub():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}-{j} = {i-j}")
        print()

def table_mul():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}*{j} = {i*j}")
        print()

def table_div():
    for i in range(1, 10):
        for j in range(1, 10):
            print(f"{i}/{j} = {i/j}")
        print()

choice = input()

if choice == "table_sum":
    table_sum()
elif choice == "table_sub":
    table_sub()
elif choice == "table_mul":
    table_mul()
elif choice == "table_div":
    table_div()
```
