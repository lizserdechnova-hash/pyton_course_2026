```python
import random

# вектор
def create_vector(n):
    vector = []
    for i in range(n):
        vector.append(random.random())
    return vector

# матрица
def create_matrix(m, n):
    matrix = []
    for i in range(m):
        row = create_vector(n)
        matrix.append(row)
    return matrix

# матрица на вектор
def multiply_matrix_vector(matrix, vector):
    m = len(matrix)
    n = len(vector)

    result = []
    for i in range(m):
        sum_val = 0
        for j in range(n):
            sum_val += matrix[i][j] * vector[j]
        result.append(sum_val)
    return result

# вывод матрицы
def print_matrix(matrix):
    for row in matrix:
        for val in row:
            print(f"{val:.2f}", end=" ")
        print()

# вывод вектора
def print_vector(vector):
    for val in vector:
        print(f"{val:.2f}", end=" ")
    print()

# сумма диагональных элементов
def sum_diagonal(matrix):
    total = 0
    size = min(len(matrix), len(matrix[0]))
    for i in range(size):
        total += matrix[i][i]
    return total

# двумерная свертка
def convolution_2d(image, kernel):
    h = len(image)
    w = len(image[0])
    kh = len(kernel)
    kw = len(kernel[0])

    pad_h = kh // 2
    pad_w = kw // 2

    result = []
    for i in range(h):
        row = []
        for j in range(w):
            total = 0
            for ki in range(kh):
                for kj in range(kw):
                    ii = i + ki - pad_h
                    jj = j + kj - pad_w
                    if 0 <= ii < h and 0 <= jj < w:
                        total += image[ii][jj] * kernel[ki][kj]
            row.append(total)
        result.append(row)
    return result



image = create_matrix(5, 5)
vector = create_vector(5)
kernel = create_matrix(3, 3)

print_matrix(image)
print()

print_vector(vector)
print()

print_matrix(kernel)
print()

result_vector = multiply_matrix_vector(image, vector)
print()

print_vector(result_vector)
print()

diag_sum = sum_diagonal(image)
print(diag_sum)
print()

convolved_image = convolution_2d(image, kernel)
print_matrix(convolved_image)
```
