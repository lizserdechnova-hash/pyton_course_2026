```python
# из RGB в YIQ
def rgb_to_yiq(v):
    r, g, b = v[0], v[1], v[2]
    y = 0.299 * r + 0.587 * g + 0.114 * b
    i = 0.5959 * r - 0.2746 * g - 0.3213 * b
    q = 0.2115 * r - 0.5227 * g + 0.3112 * b
    return [y, i, q, 1]


# из YIQ в RGB
def yiq_to_rgb(v):
    y, i, q = v[0], v[1], v[2]
    r = y + 0.956 * i + 0.619 * q
    g = y - 0.272 * i - 0.647 * q
    b = y - 1.106 * i + 1.703 * q
    return [r, g, b, 0]


def convert_color(v):
    if v[3] == 0:
        return rgb_to_yiq(v)
    else:
        return yiq_to_rgb(v)


# пример
color_rgb = [120, 200, 50, 0]
print("исходный вектор:", color_rgb)

color_yiq = convert_color(color_rgb)
print("после конвертации:", [round(x, 2) for x in color_yiq])

color_back = convert_color(color_yiq)
print("обратно:", [round(x, 2) for x in color_back])
```
